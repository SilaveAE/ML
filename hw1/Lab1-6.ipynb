4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
a = np.array([1, 1, 1, 1, 2 ,34, 5, 6, 7, 8, 9, 34, 11, 23, 42, 42, 42])
b = np.array([2, 2, 2, 2, 2, 2, 45, 65, 42, 34, 52, 12, 12 , 34])

unique1, cnt1 = np.unique(a, return_counts=True)
unique2, cnt2 = np.unique(b, return_counts=True)

if cnt1.max() > cnt2.max():
    print(f"Самое часто встречаемое значение в первом массиве: {unique1[np.argmax(cnt1)]}")
elif cnt1.max() < cnt2.max():
    print(f"Самое часто встречаемое значение во втором массиве: {unique2[np.argmax(cnt2)]}")
else:
    print(f"Самое часто встречаемое значение и в первом и во втором массиве, I: {unique1[np.argmax(cnt1)]}, II: {unique2[np.argmax(cnt2)]}")
````

````python
a = np.random.randint(0, 100, size=(10, 2))

x = a[:, 0]
y = a[:, 1]

r = np.sqrt(x**2 + y**2)

theta_rad = np.arctan2(y, x)
theta_pol = np.degrees(theta_rad)

print(np.column_stack([r, theta_pol]))
````

````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv('student_perfomance.csv')
df.head()
````

1. Посчитайте среднее кол-во часов уделяемое на сон студентами

````python
print(df['Sleep_Hours'].mean())
````

2. Сколько процентов студентов посещают внеклассовые активности?

````python
extra_activity_percent = df['Extracurricular_Activities'].value_counts(["Yes"]) * 100
print(f'Студентов, посещающих доп. занятия: {extra_activity_percent["Yes"]}%')
````

3. Правда ли, что в среднем девушки сдают экзамены лучше, чем мужчины

````python
exam_score_female = df[df['Gender'] == 'Female']['Exam_Score'].mean()
exam_score_male = df[df['Gender'] == 'Male']['Exam_Score'].mean()
if exam_score_female > exam_score_male:
    print("Девушки в среднем сдают экзамен лучше")
else:
    print("Мужчины сдают лучше")
````

4. Выведите 25 и 75 перцентили времени сна студентов

````python
q25 = df['Sleep_Hours'].quantile(0.25)
q75 = df['Sleep_Hours'].quantile(0.75)
print(f"25 перцентили сна: {q25}\n75 перцентили сна: {q75}")
````

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
sleep_hours = df['Sleep_Hours']
score = df[sleep_hours >= 75]['Exam_Score'].mean() > df[sleep_hours <= q25]["Exam_Score"].mean()
print("Те, кто спят больше сдают лучше" if score else "Те, кто спят меньше сдают лучше")
````

6. Влияет ли физическая активность на результаты экзамена?

````python
phy_activity = df['Physical_Activity']
if(df[phy_activity > 3]['Exam_Score'].mean() > df[phy_activity <= 3]['Exam_Score'].mean()):
    print("Физическая активность влияет на сдачу экзамена")
else:
    print("Физическая активность не влияет на сдачу экзамена")
````

7. Правда ли, что слуденты, чьи родители имеют высшее образование(College || Postgraduate), сдают экзамены лучше, чем остальные?

````python
parents = df['Parental_Education_Level']
if(df[parents == 'High School']['Exam_Score'].mean() < df[parents != 'High School']['Exam_Score'].mean()):
    print("Студенты, у которых родители с высшим образованием, сдают экзамены лучше")
else:
    print("Образование не влияет")
````

8. Как влияет участие родителей на результаты экзаменов

````python
parental_inv = df['Parental_Involvement']
if(df[parental_inv != "Low"]['Exam_Score'].mean() > df[parental_inv == 'Low']['Exam_Score'].mean()):
    print("Студенты, у которых родители больше вовлечены в образование ребенка, сдают экзамены лучше")
else:
    print("Не влияет")
````

9. Правда ли, что в семьях с доходом больше среднего дети сдают экзамены лучше?

````python
familly_inc = df["Family_Income"]
print("В семьях с доходом выше среднего экзамены сдают лучше" if df[familly_inc == 'High']['Exam_Score'].mean() > df[familly_inc != "High"]["Exam_Score"].mean() else "Доход семьи не влияет на успеваемость")
````

10. Посчитайте долю студентов сдавших экзамен на более чем 70 баллов в семьях с низким доходом и отсутствием доступа к интернету

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

1. Правда ли, что количество часов затрачиваемых на сон распределено примерно нормально?

````python
df = pd.read_csv('student_perfomance.csv')
sleep_h = df['Sleep_Hours']

plt.hist(df['Sleep_Hours'], bins=15, density=True, alpha=0.6, color='b')

mu = sleep_h.mean()
sigma = sleep_h.std()

x = np.linspace(sleep_h.min(), sleep_h.max())
y = (1/(sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu)/sigma)**2)

plt.plot(x, y, 'g-', linewidth=2, label='Нормальное распределение')
plt.title('Распределение часов сна')
plt.xlabel('Часы сна')
plt.ylabel('Плотность')
plt.legend()
plt.show()
````

2. Постройте гистограмму распределения количества часов затрачиваемых на обучение среди мальчиков и девочек

````python
gender = df['Gender']
plt.hist(df[gender == 'Male']['Hours_Studied'], alpha=0.7, color='purple', edgecolor='black', label='Мужчины')
plt.hist(df[gender == 'Female']['Hours_Studied'], alpha=0.7, color='brown', edgecolor='black', label='Девушки')

plt.legend()
plt.xlabel('Количество часов')
plt.ylabel('Плотность студентов')
plt.show()
````

3. Как зависит результат конечного экзамена от результата первого экзамена

````python
prev_score = df['Previous_Scores']
final_score = df['Exam_Score']

plt.scatter(prev_score, final_score, alpha=0.7)
plt.title('Зависимость конечного экзамена от первого')
plt.show()

print(f"Корреляция: {prev_score.corr(final_score)}")
````
